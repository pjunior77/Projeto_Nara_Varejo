```python
#Importo as bibliotecas
import pandas as pd
```

```python
#Carrego os dados dos atendimentos
dfAtendimentos = pd.read_csv("dfAtendimentos.csv", delimiter=",")
dfAtendimentos.head(10)
```

```python
#Guardo o número total de atendimentos
totalAtendimentos = dfAtendimentos.__len__()
totalAtendimentos
```

```python
#Seleciono os atendimentos eficientes (tempo espera <= 60 segundos)
dfAtendEficientes = dfAtendimentos.loc[dfAtendimentos["Tempo_Resposta"] <= 60]
dfAtendEficientes
```

```python
#Guardo o número total de atendimentos eficientes
totalAtendimentosEficientes = dfAtendEficientes.__len__()
totalAtendimentosEficientes
```

```python
#Calculo o percentual de atendimentos eficientes em relação ao total
percentAtenEficientes = (totalAtendimentosEficientes / totalAtendimentos) * 100
print(f'O percentual de atendimentos eficientes foi de: {percentAtenEficientes.__round__(2)} %')
```

```python
#Relação dos atendimentos eficientes com a satisfação
dfSatisfacaoAtendEficientes = dfAtendEficientes.groupby("Nota_Satisfacao")["ID_Atendimento"].count().reset_index()
dfSatisfacaoAtendEficientes
```
