```python
#Importo as bibliotecas
import pandas as pd
```

```python
#Carrega a base de clientes
dfClientes = pd.read_csv("clientes.csv", delimiter=",")
dfClientes.head(2)
```

```python
#Crio uma coluna para incluir o perfil do cliente
dfClientes["Perfil"] = ""
dfClientes.head(2)
```

```python
#Atualizo o perfildo cliente conforme a sua idade
dfClientes.loc[(dfClientes["Idade"] >= 0) & (dfClientes["Idade"] < 23), "Perfil"] = "Jovem"
dfClientes.loc[(dfClientes["Idade"] >= 23) & (dfClientes["Idade"] <= 65), "Perfil"] = "Adulto"
dfClientes.loc[(dfClientes["Idade"] > 65), "Perfil"] = "Idoso"

dfClientes
```

```python
#Exporto os dados para um arquivo CSV
dfClientes.to_csv("dfClientes.csv", sep=",", index=False)
```
