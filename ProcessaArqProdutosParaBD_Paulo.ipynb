```python
#Importo as bibliotecas
import pandas as pd
```

```python
#Carrega as bases de produtos
dfProd = pd.read_csv("produtos.csv", delimiter=",")
dfProdMarca = pd.read_csv("produtos_sem_marca_corrigidos.csv", delimiter=",")
dfProdAdicional = pd.read_csv("Produto adicional.csv", delimiter=";")
```

```python
#Apago a coluna "Marca" porque quado fazia o merge apareciam as colunas Marca_X e Marca_Y
dfProd = dfProd.drop(columns=["Marca"])
dfProd
```

```python
#Realizo o merge entre os dataFrames para pegar a coluna "Marca"
dfProdMerge = pd.merge(dfProd, dfProdMarca, left_on="ID_Produto", right_on="ID_Produto", how="inner")
dfProdMerge
```

```python
#Incluo um valor 0.00 na coluna preço, porque estava vazio
dfProdAdicional["Preco"] = 0.00
dfProdAdicional
```

```python
#Concateno as bases em uma só
dfProdutos = pd.concat([dfProdMerge, dfProdAdicional], ignore_index=True)
dfProdutos
```

```python
#Exporto os dados para um arquivo CSV
dfProdutos.to_csv("dfProdutos.csv", sep=",", index=False)
```
